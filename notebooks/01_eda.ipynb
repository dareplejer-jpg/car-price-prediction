# 01. Istrazivacka analiza podataka (EDA) - Cene polovnih automobila

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv('../data/cars.csv')
print('Dimenzije:', df.shape)
df.head()
```

```python
print('Nedostajuce vrednosti:')
print(df.isna().sum())
```

```python
print('Deskriptivna statistika cene:')
print(df['priceUSD'].describe())
```

```python
plt.figure(figsize=(10, 4))
sns.histplot(df['priceUSD'], kde=True, bins=30)
plt.title('Raspodela cene automobila')
plt.xlabel('Cena (USD)')
plt.show()
```
